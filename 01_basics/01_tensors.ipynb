# PyTorch Tensors
Tensors are the core data structure in PyTorch — like NumPy arrays but GPU-capable and autograd-aware.

```python
import torch
import numpy as np
```

## 1. Creating Tensors

```python
# From Python list
t = torch.tensor([1.0, 2.0, 3.0])
print(t)

# From nested list (2D)
m = torch.tensor([[1, 2], [3, 4]])
print(m)
```

```python
# Factory functions
print(torch.zeros(3, 4))       # all zeros
print(torch.ones(2, 3))        # all ones
print(torch.rand(2, 3))        # uniform [0, 1)
print(torch.randn(2, 3))       # standard normal
print(torch.arange(0, 10, 2))  # like range()
print(torch.eye(3))            # identity matrix
```

## 2. Tensor Attributes

```python
x = torch.randn(3, 4)

print("shape:", x.shape)      # torch.Size([3, 4])
print("dtype:", x.dtype)      # torch.float32
print("device:", x.device)    # cpu
print("ndim:", x.ndim)        # 2
print("numel:", x.numel())    # 12 — total elements
```

```python
# Dtype casting
x_int = torch.tensor([1, 2, 3])           # int64 by default
x_float = x_int.float()                   # → float32
x_double = x_int.to(torch.float64)        # → float64
print(x_int.dtype, x_float.dtype, x_double.dtype)
```

## 3. Indexing and Slicing

```python
x = torch.arange(12).reshape(3, 4)
print(x)

print(x[0])        # first row
print(x[:, 1])     # second column
print(x[1, 2])     # scalar at row=1, col=2
print(x[0:2, 1:3]) # submatrix
```

```python
# Boolean masking
x = torch.tensor([1, -2, 3, -4, 5])
mask = x > 0
print(mask)       # [True, False, True, False, True]
print(x[mask])    # [1, 3, 5]
```

## 4. Reshaping

```python
x = torch.arange(12)

a = x.reshape(3, 4)     # reshape — may copy
b = x.view(3, 4)        # view — shares memory, requires contiguous
c = x.reshape(3, -1)    # -1 = infer dimension

print(a.shape, b.shape, c.shape)
```

```python
# squeeze / unsqueeze — add or remove size-1 dims
x = torch.randn(3, 1, 4)
print(x.squeeze().shape)         # (3, 4) — removes size-1
print(x.unsqueeze(0).shape)      # (1, 3, 1, 4) — adds dim at position 0
```

## 5. Math Operations

```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print(a + b)          # element-wise add
print(a * b)          # element-wise multiply
print(torch.dot(a, b))  # dot product → scalar
print(a.sum())
print(a.mean())
print(a.max())
```

```python
# Matrix multiplication
A = torch.randn(3, 4)
B = torch.randn(4, 5)

C = A @ B             # preferred syntax
D = torch.matmul(A, B)
print(C.shape)        # (3, 5)
```

```python
# Broadcasting — smaller tensor expands to match larger
x = torch.ones(3, 4)
bias = torch.tensor([1.0, 2.0, 3.0, 4.0])  # shape (4,)
print((x + bias).shape)   # (3, 4) — bias broadcast across rows
```

## 6. NumPy Interop

```python
# Tensor → NumPy (shares memory on CPU)
t = torch.ones(3)
n = t.numpy()
print(n)

# NumPy → Tensor
n2 = np.array([1.0, 2.0, 3.0])
t2 = torch.from_numpy(n2)
print(t2)
```

## 7. GPU (if available)

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print("Using:", device)

x = torch.randn(3, 4).to(device)
print(x.device)
```
